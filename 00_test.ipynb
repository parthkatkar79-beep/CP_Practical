```python
#gate marks of student  and give grade based on marks
#above 90 - A , above 80 - B , above 70 - C , above 60 - D , below 60 - F
marks = int(input("Enter the marks of the student: "))
if marks > 90:
    print("Grade: A")
elif marks > 80:
    print("Grade: B")
elif marks > 70:
    print("Grade: C")
elif marks > 60:
    print("Grade: D")
else:   
    print("Grade: F")
```
